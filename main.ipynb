# Penanganan Anomali


## Baca ulang data

```python
df = pd.read_csv("online_retail_data.csv", dtype={"Invoice": str, "StockCode": str})
df["Baris CSV"] = df.index + 2
kolom_asli = ["Invoice", "StockCode", "Description", "Quantity",
              "InvoiceDate", "Price", "Customer ID", "Country"]
print("Jumlah Baris Awal:", f"{len(df):,}")
```

## Hapus baris duplikat

```python
df = df.drop_duplicates(subset=kolom_asli).copy()
print("Setelah buang duplikat:", f"{len(df):,}")
```

## Rapihkan format

```python
df["StockCode"] = df["StockCode"].str.strip().str.upper()
df["Description"] = (df["Description"].str.strip()
                                      .str.replace(r"\s+", " ", regex=True)
                                      .str.upper())
df["Country"] = df["Country"].str.strip().replace({"EIRE": "Ireland",
                                                   "RSA": "South Africa"})
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Customer ID"] = df["Customer ID"].astype("Int64")
```

## Keluarkan stock tanpa CustomerID

```python
adj = (df["Price"] == 0) & df["Customer ID"].isna()
keluar_stok = df[adj].copy()
keluar_stok["alasan"] = "Perubahan stok (Price 0, tanpa customer)"
df = df[~adj].copy()
print("Dikeluarkan (perubahan stok):", f"{len(keluar_stok):,}")
```

## Buang invoice A, invoice C dengan quantity, dan Positif dan data test

```python
is_a    = df["Invoice"].str.startswith("A")
is_test = df["StockCode"].str.startswith("TEST")
is_cpos = df["Invoice"].str.startswith("C") & (df["Quantity"] > 0)
buang = is_a | is_test | is_cpos

keluar_lain = df[buang].copy()
keluar_lain["alasan"] = np.select(
    [is_a[buang], is_test[buang], is_cpos[buang]],
    ["Invoice A (adjust bad debt)", "Data TEST", "Invoice C dengan Quantity positif"],
    default="Lainnya")
df = df[~buang].copy()
print("Dikeluarkan (A, TEST, dan C positif):", f"{len(keluar_lain):,}")
```

### Tandai jenis invoice dan jenis produk

```python
df["invoice_type"] = np.where(df["Invoice"].str.startswith("C"), "RETURN", "SALE")

non_produk = ["POST", "DOT", "M", "C2", "D", "S", "BANK CHARGES",
              "ADJUST", "ADJUST2", "AMAZONFEE", "CRUK"]
df["product_type"] = np.where(df["StockCode"].isin(non_produk)
                              | df["StockCode"].str.startswith("GIFT_"),
                              "NON_PRODUCT", "PRODUCT")
print(df["invoice_type"].value_counts())
print(df["product_type"].value_counts())
```

## Satu StockCode satu nama, satu invoice satu waktu 

```python
nama_utama = df.groupby("StockCode")["Description"].agg(lambda s: s.mode()[0])
df["Description"] = df["StockCode"].map(nama_utama)
df["InvoiceDate"] = df.groupby("Invoice")["InvoiceDate"].transform("min")
```

## Validasi

```python
print("Baris akhir:", f"{len(df):,}")
print("StockCode dengan >1 nama:", (df.groupby("StockCode")["Description"].nunique() > 1).sum())
print("Invoice dengan >1 timestamp:", (df.groupby("Invoice")["InvoiceDate"].nunique() > 1).sum())
print("Description kosong:", df["Description"].isna().sum())
print("Price negatif:", (df["Price"] < 0).sum())
print("StockCode huruf kecil:", df["StockCode"].str.contains(r"[a-z]").sum())
print("Invoice A:", df["Invoice"].str.startswith("A").sum())
```

## Simpan di file baru

```python
df.drop(columns="Baris CSV").to_csv("clean_transactions.csv", index=False)
removed = pd.concat([keluar_stok, keluar_lain])
removed.to_csv("removed_rows.csv", index=False)
print("\nTersimpan: clean_transactions.csv dan removed_rows.csv")
print(removed["alasan"].value_counts())
```
